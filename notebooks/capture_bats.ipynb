```python
import torch
import torchvision.models as models

# Load pre-trained ResNet-50 model
model = models.resnet50(pretrained=True)
model.eval()

# Specify the BatchNorm layer to target
target_layer_name = 'layer4.2.bn3'

# Dictionary to store BatchNorm statistics
bn_stats = {}

# Retrieve the specific BatchNorm layer
target_layer = dict(model.named_modules())[target_layer_name]

if isinstance(target_layer, torch.nn.BatchNorm2d):
    # Compute the overall mean and std (single value)
    overall_mean = target_layer.running_mean.mean().item()
    overall_std = target_layer.running_var.sqrt().mean().item()  # Convert variance to std

    # Save to separate .pt files
    torch.save(overall_mean, 'batchnorm_overall_mean.pt')
    torch.save(overall_std, 'batchnorm_overall_std.pt')

    print("Overall BatchNorm statistics saved!")
    print(f"Overall Mean: {overall_mean}")
    print(f"Overall Std: {overall_std}")
else:
    print(f"Layer {target_layer_name} is not a BatchNorm2d layer!")
```
