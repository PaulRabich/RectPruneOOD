```python
def plot_table(
    id_dataset="imagenet",
    ood_dataset=ood_datasets,
    rectification_method="",
    prune_method="",
    metric="energy_score",
):
    df = pd.read_csv(os.path.join(get_top_level_path(), "results", 'scores', 'collected_results.csv'),keep_default_na=False) 
    results = df.loc[
        (df["id_dataset"] == id_dataset)
        & (df["rectification_method"] == rectification_method)
        & (df["prune_method"] == prune_method)
        & (df["score_metric"] == metric),
        ["ood_dataset", "FPR", "AUROC"],
    ]
    
    results["FPR"] = results["FPR"] * 100
    results["AUROC"] = results["AUROC"] * 100

    rectification_method = "None" if rectification_method == "" else rectification_method
    prune_method = "None" if prune_method == "" else prune_method

    print("ID_DATASET: ", id_dataset)
    print("RECTIFICATION_METHOD: ", rectification_method)
    print("PRUNE_METHOD: ", prune_method)
    print("METRIC: ", metric)
    print()
    
    # print the resutls in a nice table
    # Determine the maximum width for the ood_dataset column
    max_width = max(len("ood_datasets"), max(len(str(val)) for val in results["ood_dataset"])) + 2

    print(f"{'ood_datasets':<{max_width}} {'FPR':<8} {'AUROC':<8}")
    print("-" * (max_width + 16))

    # Print rows with adjusted widths
    for i in range(len(results)):
        print(
            f"{results.iloc[i]['ood_dataset']:<{max_width}} "
            f"{results.iloc[i]['FPR']:<8.2f} "
            f"{results.iloc[i]['AUROC']:<8.2f}"
        )



plot_table(result_df, id_dataset="imagenet_full_spec", rectification_method="", prune_method="", metric="energy_score")
```
